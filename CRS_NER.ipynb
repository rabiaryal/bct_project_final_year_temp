```python
import pandas as pd
import spacy
from tqdm.notebook import tqdm
from spacy.pipeline import EntityRuler
from spacy.tokens import DocBin
import os

nlp = spacy.blank("en")

# Adding EntityRuler to the pipeline
if "ner" not in nlp.pipe_names:
    ner = nlp.add_pipe("ner")

ruler = nlp.add_pipe("entity_ruler", before="ner")


patterns = [

    # 1 lakh per semester
    {"label": "Fee", "pattern": [
        {"LIKE_NUM": True},
        {"LOWER": {"IN": ["lakh", "lakhs", "lac"]}},
        {"LOWER": "per"},
        {"LOWER": "semester"}
    ]},

    # 1 lakh per year
    {"label": "Fee", "pattern": [
        {"LIKE_NUM": True},
        {"LOWER": {"IN": ["lakh", "lakhs", "lac"]}},
        {"LOWER": "per"},
        {"LOWER": "year"}
    ]},

    # Rs 450000 / NPR 500000
    {"label": "Fee", "pattern": [
        {"LOWER": {"IN": ["rs", "npr"]}},
        {"LIKE_NUM": True}
    ]},

    # 4.5 lakh / lac
    {"label": "Fee", "pattern": [
        {"LIKE_NUM": True},
        {"LOWER": {"IN": ["lakh", "lakhs", "lac"]}}
    ]},

    # 90 thousand
    {"label": "Fee", "pattern": [
        {"LIKE_NUM": True},
        {"LOWER": {"IN": ["thousand", "thousands"]}}
    ]},

    # 50000 per semester / year
    {"label": "Fee", "pattern": [
        {"LIKE_NUM": True},
        {"LOWER": "per"},
        {"LOWER": {"IN": ["semester", "year"]}}
    ]},


    # Course patterns for single word courses
    {"label": "Course", "pattern": [{"LOWER": "computer"}]},
    {"label": "Course", "pattern": [{"LOWER": "civil"}]},
    {"label": "Course", "pattern": [{"LOWER": "electrical"}]},
    {"label": "Course", "pattern": [{"LOWER": "electronics"}]},
    {"label": "Course", "pattern": [{"LOWER": "chemical"}]},
    {"label": "Course", "pattern": [{"LOWER": "aerospace"}]},
    {"label": "Course", "pattern": [{"LOWER": "automobile"}]},
    {"label": "Course", "pattern": [{"LOWER": "mechanical"}]},
    {"label": "Course", "pattern": [{"LOWER": "geomatics"}]},
    {"label": "Course", "pattern": [{"LOWER": "industrial"}]},
    {"label": "Course", "pattern": [{"LOWER": "agriculture"}]}
]

ruler.add_patterns(patterns)


db = DocBin()


for text, annotation in tqdm(train_data):
    doc = nlp.make_doc(text)
    ents = []
    for start, end, label in annotation["entities"]:
        span = doc.char_span(start, end, label=label, alignment_mode="contract")
        if span is None:
            print(f"Skipping entity in: '{text[start:end]}'")
        else:
            ents.append(span)
    doc.ents = ents
    db.add(doc)


output_path = "/content/train.spacy"


db.to_disk(output_path)
```
